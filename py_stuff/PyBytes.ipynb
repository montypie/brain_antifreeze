```python
def rhalf_tree(n):
    # draws a right side of a piramid
    for r in range(1, n+1):
        for s in range(1, r+1):
            print("*", end="")
        print()
```

```python
def full_tree(howtall):
    # draws a full piramid
    
    for r in range(1, howtall+1):
        # draw spaces
        for sp in range(0, howtall-r):
            print(" ", end="")
        # draw stars
        for st in range(1, 2*r):
            print("*", end="")
        print()
```

```python
full_tree(8)
```

```python
def is_palindrome(str_in):
    # check whether a string is a palindrome

    s = str_in.replace(" ", "")
    ml = len(s) - 1
    pal = True
    for i in range(0, ml//2+1):
        if s[i] != s[ml-i]:
            pal = False
    return pal
```

```python
def py_palindrome(str_in):
    # the same but the pythonic way

    s = str_in.replace(" ", "")
    return s == s[::-1]
```

```python
def test_pal():
    assertTrue(py_palindrome("а роза упала на лапу азора"))
    assertTrue(is_palindrome("123 67 76321"))
    assertFalse(py_palindrome("notapalindrome"))
```

```python
def numchar_tree(n, t):
    # piramid from integers or letters
    for i in range(1, n+1):
        for s in range(0, n-i):
            print(" ", end="")
        for j in list(range(1, 2*i)):
            if t == "chars":
                j = chr(j+64) # first char A is 65 in ASCII
            print(j, end="")
        print()
```

```python
import random
import string
from collections import Counter

def make_lets(n):
    # creates a list of random letters of n-size
    
    letters = string.ascii_letters.lower()
    return random.choices(letters, k=n)

def make_nums(r, n):
    # creates a list of random unique integers of n-size
    
    nums = random.sample(range(1, r), n)
    print(nums)
    return nums

def most_pop(alist):
    # finds the most frequent element in a list
    
    ctr = Counter(alist)
    res = ctr.most_common(1)
    return res[0]

def removeDuplicates(alist):
    # gets rid of duplicates in a list

    cts = Counter(alist)
    dups = [x for x, count in cts.items() if count > 1]
    # print(f"Removed: {dups}")
    clean = [x for x, count in cts.items() if count == 1]
    # print(f"Result: {clean}")
    return clean
    
```

```python
def missingNums(arr):
    # finds missing numbers in a list
    arr.sort()
    mis = []
    for i in range(0, len(arr)-1):
        if arr[i] + 1 != arr[i+1]:
            mis.append(arr[i] + 1)
    return mis
```

```python
def areAnagrams(s1, s2):
    if len(s1) == len(s2):
        sl1 = list(s1)
        sl1.sort()
        sl2 = list(s2)
        sl2.sort()
        if sl1 == sl2:
            return True
```

```python
def binarysearch(arr, t):
    # binarysearch for target, if multiple: returns the smallest index
    
    if t not in arr:
        print(f"{t} is not in array")
        return False
            
    start = 0
    end = len(arr) - 1
        
    while start <= end:
        mid = (start + end) // 2
        num = arr[mid]
        if t <= num:
            res = mid
            end = mid - 1
        else:
            start = mid + 1
    print(f"Index of {t} is {res}")
```

```python
nums = [2,3,3,3,5,7,8,9]
binarysearch(nums, 3)
```

```python
def intToRoman(n):
    
    roman_map = [
        (1000, 'M'), (900, 'CM'), (500, 'D'), (400, 'CD'),
        (100, 'C'),  (90, 'XC'), (50, 'L'),  (40, 'XL'),
        (10, 'X'),   (9, 'IX'),  (5, 'V'),   (4, 'IV'),
        (1, 'I')
    ]
    roman = ''
    for val, sym in roman_map:
        while n >= val:
            roman += sym
            n -= val
    return roman
```

```python
def romanToInt(s):
    # AI-way to achieve the roman to integer convertion w reversing s
    roman_map = {
        'I': 1,    'V': 5,    'X': 10,    'L': 50,
        'C': 100,  'D': 500,  'M': 1000
        }

    res = 0
    prev_value = 0

    for char in reversed(s.upper()):
        value = roman_map[char]
        if value < prev_value:
            res -= value
        else:
            res += value
        prev_value = value
    return res
```

```python
romanToInt('MMMMMDCXLVII')
```

```python
def roman2int(s):
    # an alternative way to achieve the roman to integer convertion (w/o reversing s)
    roman_map = [
        (900, 'CM'), (400, 'CD'), (90, 'XC'), (40, 'XL'), (9, 'IX'), (4, 'IV'),
        (1000, 'M'), (500, 'D'), (100, 'C'), (50, 'L'), (10, 'X'), (5, 'V'), (1, 'I')
        ]
    res = 0
    
    for val, sym in roman_map:
        cnt = s.count(sym)
        if cnt > 0:
            res += val*cnt
            s = s.replace(sym, "")
    return res
```

```python
def RLencode(s : str) -> str:
     # run length encoding; ex.: aaaabbcccaddd >> a4b2c3a1d3
    
    res = ""
    prev_char = ""
    count = 0
        
    for char in s:
        if char != prev_char:
            if count == 0:
                res += char
            else:
                res += str(count) + char
            prev_char = char
            count = 1
        else:
            count += 1
    res += str(count)
    return res
```

```python
RLencode('aaaabbcccadd')
```

```python
def RLencode2(s):
    encoded = []
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            encoded.append(f"{s[i - 1]}{count}")
            count = 1

    # append the last character and its count
    encoded.append(f"{s[-1]}{count}")
    return ''.join(encoded)
```

```python
def bin2dec(bs):
    # binary string to decimal convertion

    li = list(map(int, reversed(bs)))
    order = list(reversed(range(len(li))))
    res = []
    for i in order:
        res.append(li[i]*(2**i))
    # OR in 1 line (with list comprehension and enumerate):
    # res = [int(x) * (2 ** i) for i, x in enumerate(reversed(bs))]
    return sum(res)
```

```python
bin2dec("101010")
```

```python
def decToBinary(n):
        
    res = ""
    while n > 0:
        if n % 2 == 1:
            res += '1'
        else:
            res += '0'
        n = n // 2
    return res[::-1]
```

```python
def dec2bin(n):
    db = {x: 2**x for x in reversed(range(32))}

    res = ""
    for k in db:
         if n - db[k] >= 0:
            res += "1"
            n = n - db[k]
         else:
            res += "0"
    first_1 = res.find('1')
    return res[first_1:]
```

```python
def decToBinary2(n):
        
    res = ""
    i = 0
    while n > 0:
        if n % (2**(i+1)) != 0:
            res += '1'
            n = n - 2**i
        else:
            res += '0'
        i += 1
    return res[::-1]
```
